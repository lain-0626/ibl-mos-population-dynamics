```python
from pathlib import Path
import pandas as pd

# Load global region inventory

PROJECT_DIR = Path(
    r"D:\Ephys_analysis\ibl-population-dynamics"
)

GLOBAL_INVENTORY_DIR = (
    PROJECT_DIR
    / "results"
    / "global_inventory"
)

REGION_INVENTORY_PATH = (
    GLOBAL_INVENTORY_DIR
    / "global_region_inventory.csv"
)


region_df = pd.read_csv(
    REGION_INVENTORY_PATH
)

PROTOTYPE_SESSION = (
    "e349a2e7-50a3-47ca-bc45-20d1899854ec"
)

```

```python
motor_region_candidates = (
    region_df.loc[
        region_df["region_parent"]
        .astype(str)
        .str.contains(
            "motor",
            case=False,
            na=False,
        )
    ]
    ["region_parent"]
    .drop_duplicates()
    .sort_values()
    .tolist()
)

motor_region_candidates
```

```python
TARGET_REGION = "Secondary motor area"

mos_session_inventory_df = (
    region_df.loc[
        region_df["region_parent"].eq(
            TARGET_REGION
        )
    ]
    .groupby(
        "session",
        as_index=False,
    )
    .agg(
        n_all_units=(
            "n_all_units",
            "sum",
        ),
        n_good_units=(
            "n_good_units",
            "sum",
        ),
    )
    .sort_values(
        "n_good_units",
        ascending=False,
    )
    .reset_index(drop=True)
)

display(
    mos_session_inventory_df.head(20)
)
```
